### Shapiro Wilk Test: Non Parametric Test

```python
def perform_shapiro_test(df, features, group_col):
    results = []

    for feature in features:
        # Ensure the feature exists in the DataFrame
        if feature not in df.columns:
            raise ValueError(f"The feature '{feature}' is not in the DataFrame.")
        
        # Separate the data based on the group column
        group_0 = df[df[group_col] == 0][feature]
        group_1 = df[df[group_col] == 1][feature]
        
        # Perform the Shapiro-Wilk test for each group
        if len(group_0) > 0:  # Check if there are enough data points
            stat_0, p_value_0 = shapiro(group_0)
        else:
            stat_0, p_value_0 = None, None
        
        if len(group_1) > 0:  # Check if there are enough data points
            stat_1, p_value_1 = shapiro(group_1)
        else:
            stat_1, p_value_1 = None, None
        
        # Append results
        results.append({
            'Feature': feature,
            'Group': '0',
            'Shapiro_Stat': stat_0,
            'P_Value': p_value_0
        })
        
        results.append({
            'Feature': feature,
            'Group': '1',
            'Shapiro_Stat': stat_1,
            'P_Value': p_value_1
        })
    
    # Convert the results to a DataFrame
    result_df = pd.DataFrame(results)
    return result_df
```

```python
def mann_whitney_u_test(df, columns_to_format):
    results = []

    for col in columns_to_format:
        group_0 = df[df['pdc_80_flag'] == 0][col]
        group_1 = df[df['pdc_80_flag'] == 1][col]

        # Perform Mann-Whitney U test
        u_stat, p_value = mannwhitneyu(group_0, group_1, alternative='two-sided')

        results.append({'Feature': col, 'U_Statistic': u_stat, 'P_Value': p_value})

    # Create a DataFrame with the results
    result_df = pd.DataFrame(results)
    
    return result_df
```

```python
def dataFormat(df, columns_to_format):
    result = None
    
    for col in columns_to_format:
        # Group by 'pdc_80_flag' and extract unique values for each column
        col_groups = df.groupby(['pdc_80_flag'])[col].unique()
        
        # Create a DataFrame from the unique values
        col_df = pd.DataFrame(col_groups.values.tolist(), index=col_groups.index)
        
        # Transpose the DataFrame
        col_df = col_df.T
        
        # Rename columns
        col_df = col_df.rename(columns={0: f'pdc_80_flag_0', 1: f'pdc_80_flag_1'})
        
        # Fill missing values with 0
        col_df = col_df.fillna(0)
        
        if result is None:
            result = col_df
        else:
            result = pd.concat([result, col_df], axis =1)
    
    return result
```

```python
def dataformat1(df, columns_to_format):
    result = None
    
    for col in columns_to_format:
        group_0 = df[df['pdc_80_flag'] == 0][col]
        group_1 = df[df['pdc_80_flag'] == 1][col]
        
        # Perform levene test
        stats, p_value = levene(group_0, group_1)
        
        # Create a DataFrame to store the t test results
        levene_result = pd.DataFrame({
            'category': [col],
            'P_Value': [p_value]
        })
        
        if result is None:
            result = levene_result
        else:
            result = pd.concat([result, levene_result])
            
    return result
```

### p_value <= 0.05. We failed to reject the null hypothesis that pdc_80_flag_0 and pdc_80_flag_1 for the features that have the same variance. We will use Welch T test instead of Student's T test.

```python
def tTestdf_difVar(df, columns_to_format):

    result = None
    
    for col in columns_to_format:
        group_0 = df[df['pdc_80_flag'] == 0][col]
        group_1 = df[df['pdc_80_flag'] == 1][col]
        
        # Perform a Welch's T test with different variances
        t_stats, p_value = stats.ttest_ind(group_0, group_1, equal_var= False)
        
        # Create a DataFrame to store the t test results
        t_test_result = pd.DataFrame({
            'category': [col],
            'P_Value': [p_value]
        })
        
        if result is None:
            result = t_test_result
        else:
            result = pd.concat([result, t_test_result])
            
    return result
```

```python
# Convert 'P_Value' column to numeric
levene_test_result['P_Value'] = pd.to_numeric(levene_test_result['P_Value'], errors='coerce')
```

```python
levene_test_stats_significant = levene_test_result[levene_test_result['P_Value'] <=0.05]
```

```python
tTest_cols = levene_test_stats_significant.set_index(['category']).index
tTest_result = tTestdf_difVar(AD_continuous, tTest_cols)
```

### p_value > 0.05. We do not reject the null hypothesis that pdc_80_flag_0 and pdc_80_flag_1 for the features that have the same variance. We will use Student's T test.

```python
levene_test_not_significant = levene_test_result[levene_test_result['P_Value'] >0.05]
```

```python
def tTestdf_sameVar(df, columns_to_format):
    result = None
    
    for col in columns_to_format:
        group_0 = df[df['pdc_80_flag'] == 0][col]
        group_1 = df[df['pdc_80_flag'] == 1][col]
        
        # Perform a T test with same variances
        t_stats, p_value = stats.ttest_ind(group_0, group_1, equal_var= True)
        
        # Create a DataFrame to store the t test results
        t_test_result = pd.DataFrame({
            'category': [col],
            'P_Value': [p_value]
        })
        
        if result is None:
            result = t_test_result
        else:
            result = pd.concat([result, t_test_result])
            
    return result
```

```python
tTest_cols_2 = levene_test_not_significant.set_index(['category']).index
tTest_result_same = tTestdf_sameVar(AD_continuous, tTest_cols_2)
```

```python

def calculate_smd(df, group_col, covariates):
    """
    Compute SMD for each covariate in a matched cohort dataset.

    Parameters:
    df (pd.DataFrame): The dataset containing the covariates and group label.
    group_col (str): The column name indicating treatment group (0 = control, 1 = treated).
    covariates (list): List of covariate column names.

    Returns:
    pd.DataFrame: SMD values for each covariate.
    """
    smd_results = {}

    # Split dataset into treated and control groups
    treated = df[df[group_col] == 1]
    control = df[df[group_col] == 0]

    for covariate in covariates:
        mean_treated = treated[covariate].mean()
        mean_control = control[covariate].mean()
        std_treated = treated[covariate].std()
        std_control = control[covariate].std()

        # Compute pooled standard deviation
        pooled_std = ((std_treated**2 + std_control**2) / 2)** 0.5

        # Compute SMD
        smd = abs(mean_treated - mean_control) / pooled_std
        smd_results[covariate] = smd

    return pd.DataFrame.from_dict(smd_results, orient='index', columns=['SMD']).sort_values(by='SMD', ascending=False)

# Example usage:
# Assuming `AC` is your dataset and `group` is the column indicating treatment/control (1/0)
# covariates_list = ['age', 'hypertension', 'smoking', 'dyslipidemia']
# smd_df = calculate_smd(AC, 'group', covariates_list)
# print(smd_df)
```

```python
smd_df = calculate_smd(AC, 'pdc_80_flag', X)
print(smd_df)
```

While most covariates achieved acceptable balance (SMD < 0.2), regionN remained highly imbalanced (SMD > 0.9) due to a lack of common support. To mitigate this, regionN was excluded from balance diagnostics but retained in the outcome model to adjust for residual confounding.

```python
# Avoid divide-by-zero with clipping
ps = AD_clean['propensity_score'].clip(1e-5, 1 - 1e-5)
logit_ps = np.log(ps / (1 - ps))
AD_clean.loc[:, 'logit_ps'] = logit_ps

caliper = 0.2 * np.std(AD_clean['logit_ps'])

```

```python
treated = AD_clean[AD_clean['pdc_80_flag'] == 1].copy()
control = AD_clean[AD_clean['pdc_80_flag'] == 0].copy()

matched_control_indices = []
matched_treated_indices = []

for i, row in treated.iterrows():
    # Compute distance between treated unit and all controls
    distances = np.abs(row['logit_ps'] - control['logit_ps'])

    # Apply caliper
    within_caliper = distances[distances <= caliper]
    
    if within_caliper.empty:
        continue  # No match within caliper

    # Get closest match
    matched_idx = within_caliper.idxmin()
    matched_control_indices.append(matched_idx)
    matched_treated_indices.append(i)

    # Remove matched control to avoid reuse
    control = control.drop(index=matched_idx)

# Build matched dataset
matched_treated = AD_clean.loc[matched_treated_indices]
matched_control = AD_clean.loc[matched_control_indices]
matched_data = pd.concat([matched_treated, matched_control], axis=0)
```
